```python
import pandas as pd
```

```python
# Load the dataset
population_file_path = "population_USA.csv"  # Update this with your actual file path
population_df = pd.read_csv(population_file_path)
```

```python
population_df.head()
```

```python
# Keep only the row where NAME is "United States"
population_usa_df = population_df[population_df["NAME"] == "United States"]
```

```python
# Pivot the data to get 'Year' and 'Population' as columns
population_long_df = pd.melt(
    population_usa_df,
    id_vars=["NAME"],  # Keep only NAME column
    value_vars=["POPESTIMATE2020", "POPESTIMATE2021", "POPESTIMATE2022", "POPESTIMATE2023"],
    var_name="Year",
    value_name="Population"
)
```

```python
population_long_df.head()
```

```python
# Extract the numeric year from column names
population_long_df["Year"] = population_long_df["Year"].str.extract("(\d{4})").astype(int)
```

```python
population_cleaned_df = population_long_df[["Year", "Population"]]
```

```python
population_cleaned_df.head()
```

```python
# Save the cleaned data to an Excel file
cleaned_population_file_path = "Cleaned_Population_USA.xlsx"
population_cleaned_df.to_excel(cleaned_population_file_path, index=False)
```
